Importing libraries 

```python
import pandas as pd 
import numpy as np 
import random
from faker import Faker
from datetime import datetime , timedelta

fake = Faker('en_IN')

random.seed(42)
np.random.seed(42)
```

Generate Customer Churn Dataset 

```python
customer_data = []

cities = [
    "Delhi","Mumbai","Bangalore","Chennai","Hyderabad",
    "Pune","Kolkata","Ahmedabad","Jaipur","Lucknow"
]

plans = ["Basic", "Standard", "Premium"]
contracts = ["Monthly", "Yearly", "Two Year"]
payments = ["UPI", "Credit Card", "Debit Card", "Net Banking", "Wallet"]

for customer_id in range(1, 5001):

    join_date = fake.date_between(start_date='-5y', end_date='today')

    tenure_months = random.randint(1, 60)

    plan_type = random.choices(
        plans,
        weights=[40, 40, 20]
    )[0]

    if plan_type == "Basic":
        monthly_charges = random.randint(300, 700)

    elif plan_type == "Standard":
        monthly_charges = random.randint(700, 1200)

    else:
        monthly_charges = random.randint(1200, 2500)

    contract_type = random.choices(
        contracts,
        weights=[50, 30, 20]
    )[0]

    support_tickets = random.randint(0, 15)

    satisfaction_score = round(
        random.uniform(1.0, 5.0),
        1
    )

    total_charges = round(
        monthly_charges * tenure_months,
        2
    )

    churn_probability = 0

    if contract_type == "Monthly":
        churn_probability += 0.25

    if support_tickets > 8:
        churn_probability += 0.25

    if satisfaction_score < 3:
        churn_probability += 0.30

    if tenure_months < 12:
        churn_probability += 0.15

    churn = np.random.choice(
        ["Yes", "No"],
        p=[
            min(churn_probability, 0.80),
            1 - min(churn_probability, 0.80)
        ]
    )

    customer_data.append([
        customer_id,
        fake.name(),
        random.choice(["Male", "Female"]),
        random.randint(18, 65),
        random.choice(cities),
        join_date,
        tenure_months,
        plan_type,
        monthly_charges,
        total_charges,
        contract_type,
        random.choice(payments),
        support_tickets,
        satisfaction_score,
        churn
    ])

customers_df = pd.DataFrame(
    customer_data,
    columns=[
        "customer_id",
        "customer_name",
        "gender",
        "age",
        "city",
        "join_date",
        "tenure_months",
        "plan_type",
        "monthly_charges",
        "total_charges",
        "contract_type",
        "payment_method",
        "support_tickets",
        "satisfaction_score",
        "churn"
    ]
)

customers_df.head()
```

Export DataSet 

```python
customers_df.to_csv(
    "customers.csv",
    index=False
)
print("customers.csv saved successfully")
```
